dictonaries are unordered collection they store the data in key value pairs where keys must be unique and immutable (eg:tuples,string,numbers) and values should be of any type

```python
dict1={}
print(dict1)  # This will print an empty dictionary
print(type(dict1))  # This will print <class 'dict'> indicating that dict1 is a dictionary
```

```python
dict2=dict()
print(dict2)  # This will also print an empty dictionary
print(type(dict2))  # This will also print <class 'dict'> indicating that dict2 is a dictionary
```

```python
student = {"name": "Prakash", "name2": "kumar", "name5": "raju", "name4": "hi"}
grades={1: "A", 2: "B", 3: "C", 4: "D"}


#accessing values in a dictionary
print(student["name"])


print(student["name2"])  # This will print the value associated with the key 'name2' in the student dictionary

print(student["name6"]) # This will raise a KeyError since 'name6' does not exist in the dictionary
##accessing values using get method
print(student.get("name2"))  # This will also print the value associated with the key 'name2' in the student dictionary
print(student.get("name6"))  # This will return None since 'name6' does not exist in the dictionary
print(student.get("name5", "hello"))  # This will return "hello" since 'name5' does not exist in the dictionary
print(list(student.values()).index("hi"))  # This will return the index of the value "hi" in the list of values


print(student.get("name3", "not found"))  # This will return the value associated with 'name3' or "not found" if 'name3' does not exist
```

dictonaries are mutable you can add,update,delete elements but key should be unique

```python
student["name2"] = "kumar"  # This will update the value associated with the key 'name2' to "kumar"
print(student)  # This will print the updated dictionary with 'name2' now having the value "kumar"

student["name4"] = "bhai"  # This will add a new key-value pair to the dictionary
print(student)  # This will print the updated dictionary with 'name4' now having the value "bhai"

student["name7"] = "hello"  # This will add a new key-value pair to the dictionary
print(student)  # This will print the updated dictionary with 'name5' now having the value "hello"


####using dictionary methods



student.pop("name7")  # This will remove the key-value pair with the key 'name2' from the dictionary
print(student)  # This will print the updated dictionary without 'name2'


student.popitem()  # This will remove the last inserted key-value pair from the dictionary
print(student)  # This will print the updated dictionary without the last inserted key-value pair


student.update({"name8": "teja"})  # This will update the dictionary with a new key-value pair
print(student)  # This will print the updated dictionary with 'name8' now having the value "new_value"



student.update({"name5": "raju"})  # This will update the value associated with the key 'name5' to "raju"
print(student)  # This will print the updated dictionary with 'name5' now having the value "raju"

del student["name5"]  # This will delete the key-value pair with the key 'name5' from the dictionary
print(student)  # This will print the updated dictionary without 'name5'
```

```python
print(student.keys())  # This will print all the keys in the student dictionary
print(student.values())  # This will print all the values in the student dictionary

print(student.items())  # This will print all the key-value pairs in the student dictionary in the form of list(tuples)

```

```python
for k in student.keys():
    print(k)  # This will print each key in the student dictionary
for v in student.values():
    print(v)  # This will print each value in the student dictionary

for k, v in student.items():
    print(k, v)  # This will print each key-value pair in the student dictionary
    
```

Nested Dictonaries

```python
students={"student1": {"name": "Prakash", "age": 25, "grade": "A"},
           "student2": {"name": "Kumar", "age": 22, "grade": "B"},          
              "student3": {"name": "Raju", "age": 23, "grade": "C"},}
print(students)

print(students["student1"]["name"])  # This will print the name of student1
print(students["student2"]["age"])  # This will print the age of student2
print(students["student3"]["grade"])  # This will print the grade of student3   
```

iterate over nested lists

```python
for student_id, details in students.items():
    print(f"{student_id}: {details}")  # This will print each student ID and their details
    for k,v in details.items():
        print(f"{k}: {v}")  # This will print each key-value pair in the student's details
```

```python
for x in range(5):
    print(f"{x}:{x**2}")  # This will print each number from 0 to 4 and its square
```

dict comprehension

```python
{x: x**2 for x in range(5)}  # This will create a dictionary with numbers from 0 to 4 as keys and their squares as values
```

```python
{x: x**2 for x in range(1,6)if x % 2 == 0}  # This will create a dictionary with even numbers from 0 to 4 as keys and their squares as values 
```

```python
a={"name": "Prakash", "age": 25, "grade": "A"}
b={"name": "Kumar", "age": 22, "grade": "B"}
print({**a}) 
print({**b}) # This will unpack the dictionaries a and b
print(a)

#print(a|b) # This will merge the dictionaries a and b, giving precedence to the values in b if there are any conflicts
```
