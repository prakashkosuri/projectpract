```python
print(fruits[0:3])  # Slicing the first three elements
print(fruits[2:])   # Slicing from the third element to the end
print(fruits[:4])   # Slicing from the start to the fourth element
print(fruits[-1])   # Accessing the last element
print(fruits[-3:])  # Slicing the last three elements
print(fruits[-1:-8:-1])  # Slicing from the last element to the first  element
print(fruits[::1])  # Slicing every second element
print(fruits[::-1])  # Slicing every second element
```

```python
import copy
fruits=["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(fruits)

fruits.index("banana")  # Returns the index of the first occurrence of "banana"
fruits.append("papaya")  # Adds "papaya" to the end of the list
fruits.insert(3, "grape")  # Inserts "grape" at index 2
print(fruits)

fruits.remove("kiwi")  # Removes the first occurrence of "kiwi"
print(fruits)
fruits.pop(2)  # Removes the element at index 2
print(fruits)
fruits.count("orange")  # Returns the count of "orange" in the list
fruits.reverse()
print(fruits)  # Reverses the order of the list
fruits.extend (["pear", "peach"])  # Adds multiple elements to the end of the list
print(fruits)
copied_fruits=fruits.copy()
print(fruits)  # Creates a shallow copy of the list
print(copied_fruits)  # Prints the copied list
fruits.append("kiwi")  # Adds "kiwi" to the end of the list
print(fruits)
print(copied_fruits)

deep_copy=copy.deepcopy(fruits)
print(deep_copy)  # Creates a deep copy of the list


fruits.append("papaya")  # Adds "kiwi" to the end of the list
print(fruits)
print(deep_copy)



copied_fruits.append("rice")  # Adds "kiwi" to the end of the copied list
print(copied_fruits)  # Prints the copied list with "kiwi" added

print(fruits)  # Prints the original list, which remains unchanged
```

```python
num=[[1,2,3],[4,5,6],[7,8,9]]
s_copy=copy.copy(num)  # Creates a shallow copy of the list of lists
print(s_copy)  # Prints the shallow copied list of lists
s_copy[0][0]=100  # Modifies the first element of the first sublist
print(s_copy)
print(num)  # Prints the original list of lists, which is also affected by the change in the shallow copy

s_deepcopy=copy.deepcopy(num)  # Creates a deep copy of the list of lists
print(s_deepcopy)
s_deepcopy[0][0]=200  # Modifies the first element of the first sublist in the deep copy
print(s_deepcopy)
print(num)  # Prints the original list of lists, which remains unchanged by the deep copy modification
```

```python
dict1={"name":"raju","age":20,"marks":90}
dict1.update({"address":"hyderabad"})  # Adds a new key-value pair to the dictionary
print(dict1)  # Prints the updated dictionary

dict1.pop("age")  # Removes the key "age" and its associated value from the dictionary
print(dict1)  # Prints the dictionary after removing the "age" key-value pair

dict2=dict1.fromkeys(["name","age"])  # Creates a new dictionary with specified keys and a default value
print(dict2)  # Prints the dictionary created from the specified keys

dict1.get("name")  # Retrieves the value associated with the key "name"

dict1.items()

for key, value in dict1.items():
    print(key)
    print(value)  # Iterates through the dictionary and prints each key-value pair

```
