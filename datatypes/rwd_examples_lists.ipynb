Real world examples of lists in python

create a todolist to keep track of things

```python
to_do_list=["Buy groceries", "Clean the house", "Finish homework"]
print(to_do_list)  # This will print the list of to-do items
print(type(to_do_list))  # This will print <class 'list'>
to_do_list.append("Call mom")  # This will add a new item to the end of the list
print(to_do_list)
to_do_list.remove("Clean the house")  # This will remove "Clean the house" from the list
print(to_do_list)   
if "Finish homework" in to_do_list:  # This will check if "Finish homework" is in the list
    print("You have homework to finish!")   

to_do_list.sort()  # This will sort the list in alphabetical order
print(to_do_list)

for item in to_do_list:  # This will iterate through each item in the list
    print(f"To-do item: {item}")  # This will print each to-do item
```

student grades

```python
grades = [85, 90, 78, 92, 88]
print(grades)  # This will print the list of grades

average_grade = sum(grades) / len(grades)  # This will calculate the average grade
print(f"Average grade: {average_grade:.2f}")  # This will print the average grade

maxgrades=print(f"Highest grade: {max(grades)}")  # This will print the highest grade
mingrades=print(f"Lowest grade: {min(grades)}")  # This will print the lowest grade
```

manage inventery in a store

```python
store=["Apples", "Bananas", "Oranges", "Grapes"]
store.append("Mangoes")  # This will add "Mangoes" to the store inventory
store.remove("Bananas")  # This will remove "Bananas" from the store inventory

for item in store:  # This will iterate through each item in the store inventory
    print(f"Store item: {item}")

req="dragon fruit"
```

```python
print(req)

if req in store:  # This will check if "dragon fruit" is in the store inventory
    print(f"{req} is available in the store.")
else:
    print(f"{req} is not available in the store. We can order it for you.")
```

feed back service

```python
feedback = ["Great service!", "Very helpful.", "Could be better."]

positive_feedback = sum(2 for f in feedback if "great" in f.lower() or "helpful" in f)
print("Positive feedback:", positive_feedback)  # This will print the positive feedback
```
