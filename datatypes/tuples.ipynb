tuples are ordered collection of elements similar to lists but only immutability make it different from lists

```python
a=()
print(a)  # this will print an empty tuple
print(type(a))  # This will print <class 'tuple'>
```

```python
a=(1,)
print(a)  # This will print the tuple with a single element
print(type(a))  # This will print <class 'tuple'>
```

```python
a=(1, 2, 3, 4, 5)
print(a) # This will print the tuple with multiple elements
```

```python
a=([1, 2, 3], [4, 5, 6])
print(a) # This will print the tuple with nested lists
```

```python
a=tuple([1, 2, 3])
print(a)  # This will print the tuple created from a list
```

```python
a=(1,"car",3.14,True)
print(a)  # This will print the tuple with mixed data types
```

```python
a=(1, 2, 3, 4, 5)
print(a[::-1])  # This will print the tuple in reverse order

a=(1, 2, 3, 4, 5)
print(a[1:4])  # This will print a slice of the tuple from index 1 to 3

a=(1, 2, 3, 4, 5)
print(a[2])  # This will print the element at index 2 of the tuple
```

```python
b=a.index(3)  # This will return the index of the first occurrence of the value 3 in the tuple
print(b)  # This will print the index of the value 3 in the tuple
```

```python
class1=("raju","rani","ravi")
print(class1)  # This will print the tuple with the names
print(type(class1))  # This will print <class 'tuple'>
```

```python
class2=tuple([1,"raju",3.14,True])  # This will create a tuple with mixed data types
print(class2)  # This will print the tuple with mixed data types
```

```python
class1.count("raju")  # This will count the number of occurrences of "raju" in the tuple class1
```

```python
class1.index("ravi")
```

```python
a=(1,2,3,4,5)
b=(6,7,8,9,10)
c=a+b  # This will concatenate the two tuples a and b
print(c)  # This will print the concatenated tuple
```

```python
a*3  # This will repeat the elements of the tuple a three times
```

immutablity

```python
a=[1,2,3,4,5]
a.append(6)  # This will append the value 6 to the end of the list a
print(a)  # This will print the list after appending the value


a[1]=  10  # This will change the element at index 1 of the list a to 10
print(a)  # This will print the list after changing the element at index 1
```

```python
a=(1, 2, 3, 4, 5)
a.append(6)  # This will raise an AttributeError because tuples do not have an append method
```

```python
a(1) = 10 # This will raise a TypeError because tuples are immutable and cannot be changed
```

```python
users=1,2,3,4,5
print(users)  # This will print the tuple with the user IDs
print(type(users))  # This will print <class 'tuple'>
```

```python
a,b,c,d,e= users  # This will unpack the tuple into individual variables
print(a, b, c, d, e)  # This will print the unpacked values
```

```python
a=1,2,3,4,5
p,*q,r=a  # This will unpack the tuple into p, q, and r
print(p, q, r)  # This will print the unpacked values with p as the first element, q as a list of middle elements, and r as the last element
```

```python
nested_tuple = (1, (2, 3), 4)
print(nested_tuple)
```

```python
nested_tuple[1] # This will access the second element of the tuple, which is another tuple (2, 3)
```

```python
nested_tuple[1][1]  # This will access the second element of the nested tuple, which is 3
```
