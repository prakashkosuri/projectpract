Exception is the base class or parent class every exception are derived from this class which handles all the exceptions so it will be dervived at the end of all exceptions

```python
try:
    a=int(input("enter a number"))
    b=int(input("enter a number"))
    c=a/b
except ZeroDivisionError as e:
    print(f"we encountered this error {e}")
except Exception as e:
    print(f"we encountered this error {e}")
except NameError as e: # type: ignore
    print(f"we encountered this error {e}")
else:
    print(c)
```

finally is runs no matter what.mostly we use finally to close the connection


```python
try:
    a=int(input("enter a number"))
    b=int(input("enter a number"))
    c=a/b
except ZeroDivisionError as e:
    print(f"we encountered this error {e}")
except Exception as e:
    print(f"we encountered this error {e}")
except NameError as e: # type: ignore
    print(f"we encountered this error {e}")
else:
    print(c)
finally:
    print("this will run no matter what")
```

here in below code finally is not executed Because with open(...) as f:, Python closes the file automatically when the block ends. So the manual close in finally is usually unnecessary.



```python
try:
    with open('C:/Users/Prakash kumar/Downloads/projectpract/fileoperations/sample2.txt', 'r') as f:
        content = f.readlines()
        line_count = len(content)
        word_count = sum(len(line.split()) for line in content)
        char_count = sum(len(line) for line in content)
        print(f"Lines: {line_count}, Words: {word_count}, Characters: {char_count}")
except FileNotFoundError as e:
    print(f"File not found: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
finally:
    if not f.closed:
        f.close()
        print("File closed successfully.")


```
