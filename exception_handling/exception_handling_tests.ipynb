Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.

```python
try :
    a=int(input("enter a number"))
    b=int(input("enter a number"))
    c=  a/b
except ZeroDivisionError as e:
    print(f"we encountered this error {e}")
except NameError as e:
    print(f"we encountered this error {e}")
except Exception as e:
    print(f"we encountered this error {e}")
finally:
    print(c)
```

Write a function that reads the contents of a file named `data.txt`. Use try, except, and finally blocks to handle file not found errors and ensure the file is properly closed.

```python
try :
    file = open('C:/Users/Prakash kumar/Downloads/projectpract/fileoperations/sample.txt', 'r')
    content = file.readlines()
    line_count = len(content)
    word_count = sum(len(line.split()) for line in content)
    char_count = sum(len(line) for line in content)
    print(f"Lines: {line_count}, Words: {word_count}, Characters: {char_count}")
except FileNotFoundError as e:
    print(f"File not found: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
finally:
    if not file.closed:
        file.close()
        print("File closed successfully.")
```
