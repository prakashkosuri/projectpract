```python
def factorial(n: int) -> int:   
    """    This function calculates the factorial of a number.
    :param n: The number to calculate the factorial of.
    :return: The factorial of the number.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i 
    return result

def main():
    """    Main function to demonstrate the use of factorial.
    """
    try:
        number = 100  # Example number to calculate factorial
        result = factorial(number)
        print(f"The factorial of {number} is {str(result)[:10]}")  # This will print the factorial of the number

    except ValueError as e:
        print(e)  # This will print the error message if a negative number is provided
if __name__ == "__main__":
    main()
```

```python
def prime_check(n: int) -> bool:
    """    This function checks if a number is prime.
    :param n: The number to check.
    :return: True if the number is prime, False otherwise.
    """ 
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False        
    return True # If no divisors were found, the number is prime

def main():
    """    Main function to demonstrate the use of prime_check.
    """
    number = 9
    if prime_check(number):
        print(f"{number} is a prime number.")
    else:
        print(f"{number} is not a prime number.") 

if __name__ == "__main__":
    main()     
```

```python
def keyword_arguments(**kwargs):
    """    This function demonstrates the use of keyword arguments.
    :param kwargs: Arbitrary keyword arguments.
    :return: A string representation of the keyword arguments.
    """
    for k,v in kwargs.items():
        print(f"{k} = {v}")  # This will print each keyword argument and its value
def main():
    """    Main function to demonstrate the use of keyword_arguments.
    """
    keyword_arguments(name="John", age=30, city="New York")

if __name__ == "__main__":
    main()
```

```python
##celcius to fahrenheit conversion


def celsius_to_fahrenheit(celsius: float) -> float:
    """    This function converts Celsius to Fahrenheit.
    :param celsius: The temperature in Celsius.
    :return: The temperature in Fahrenheit.
    """
    return (celsius * 9/5) + 32

def main():
    """    Main function to demonstrate the use of celsius_to_fahrenheit.
    """
    celsius = 25.0  # Example temperature in Celsius
    fahrenheit = celsius_to_fahrenheit(celsius)
    print(f"{celsius}°C is equal to {fahrenheit}°F")    

if __name__ == "__main__":
    main()
```

```python
### This function demonstrates the password strength checker

def password_strength_checker(password: str) -> str:
    """    This function checks the strength of a password.
    :param password: The password to check.
    :return: A string indicating the strength of the password.
    """
    if len(password) < 6:
        return "Weak password: Password must be at least 6 characters long."
    elif len(password) < 10:
        return "Moderate password: Password should be at least 10 characters long."
    elif len(password) < 15:
        return "Strong password: Password is strong."
    else:
        return "Very strong password: Password is very strong."
def main():
    """    Main function to demonstrate the use of password_strength_checker.
    """
    password = "mypassword1234584"  # Example password to check
    strength = password_strength_checker(password)
    print(strength)  # This will print the strength of the password
if __name__ == "__main__":
    main()  # This will call the main function to execute the code
```

```python
#palindrome checker

def is_palindrome(s: str) -> bool:
    """    This function checks if a string is a palindrome.
    :param s: The string to check.
    :return: True if the string is a palindrome, False otherwise.
    """
    s = s.lower().replace(" ", "")
    return s == s[::-1]  # Check if the string is equal to its reverse
def main():
    """    Main function to demonstrate the use of is_palindrome.
    """
    test_string = "Madam Im Adam"
    if is_palindrome(test_string):
        print(f'"{test_string}" is a palindrome.')
    else:
        print(f'"{test_string}" is not a palindrome.')

if __name__ == "__main__":
    main()
```

The Fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones, typically starting with 0 and 1. The sequence begins: 0, 1, 1, 2, 3, 5, 8, 13, 21, and so on. It's a fundamental concept in mathematics and

```python
#recussion example

def fibonacci(n: int) -> int:
    """    This function calculates the nth Fibonacci number using recursion.
    :param n: The position in the Fibonacci sequence (0-indexed).
    :return: The nth Fibonacci number.
    """ 
    if n < 0:
        raise ValueError("Fibonacci is not defined for negative numbers.")
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
    
def main():
    """    Main function to demonstrate the use of fibonacci.
    """
    try:
        n = 6  # Example position in the Fibonacci sequence
        result = fibonacci(n)
        print(f"The {n}th Fibonacci number is {result}")  # This will print the nth Fibonacci number
    except ValueError as e:
        print(e)                

if __name__ == "__main__":
    main()
```

A regular expression is a special pattern string used to match, search, or manipulate text.

2. Basic Pattern Rules
Here’s a small “cheat sheet”:

Pattern	Matches
.	Any single character (except newline)
^	Start of string
$	End of string
\d	Digit (0–9)
\D	Non-digit
\w	Word character (letters, digits, _)
\W	Non-word character
\s	Whitespace (space, tab, newline)
\S	Non-whitespace
[...]	Any one of these characters
[^...]	Any character not in brackets

3. Quantifiers (How Many Times?)
Pattern	Meaning
'*'	0 or more times
'+'	1 or more times
?	0 or 1 time
{n}	Exactly n times
{n,}	At least n times
{n,m}	Between n and m times
