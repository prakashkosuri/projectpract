filter is applied to the list or iterables it is used to filter out the items from a list or iterables based on the given condition

```python
a=[1,2,3,4,5]
list(filter(lambda x : x%2==0,a))
```

```python
def even(num):
    if num%2 ==0:
        return True
    

a=[1,2,3,4,5]

list(filter(even,a))
```

```python
a=[1,2,5,3,7,4,6]

list(filter(lambda x : x >= 5,a ))# using filter getting the numbers greater than or equal to 5 from the list
```

filter with a lambda function and multiple conditions


```python
a=[1,2,3,4,5,6,7,9,20,10,8]
list(filter(lambda x : x>5 and x%2 ==0 ,a))# using filter getting the even numbers greater than  5 from the list
```

filter operation on dictonary

```python
def get_items(people):
    return people['name']

people = [{'name':'raju','age':36},
          {'name':'manoj','age':40}]
list(map(get_items,people))
```

```python
def get_items(people):
    if people['name'] == "raju":
        return people['name'] ## here return statement is true for raju and false for manoj

people = [{'name':'raju','age':36},
          {'name':'manoj','age':40}]
list(filter(get_items,people))
```

```python
list(filter(lambda x : x['name'] == "raju",people)) #same using lambda
```
