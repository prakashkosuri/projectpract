map function applies to  all the elements in a input list or any iterable and returns a map object or an iterator

```python
a=[100,200,300]

list(map(lambda x:x+2 ,a))
```

map uses  a lazy load thechinque so we use list or tuple 

adding multiple iterable

```python
a=[1,2,3]
b=[4,5,6]

list(map(lambda x,y : x+y,a,b))
```

list of string to integers

```python
a=["1","2","3","4"]


list(map(lambda x : int(x),a))
```

```python
a=["1","2","3","4"]
list(map(int, a))
```

map of list of dictonary items

```python
def get_items(people):
    return people['name']

people = [{'name':'raju','age':36},
          {'name':'manoj','age':40}]
list(map(get_items,people))
```
