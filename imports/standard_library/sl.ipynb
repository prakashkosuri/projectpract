🔹 What is requests?

A third-party library (not built into Python by default — you install it with pip install requests).

Used to send HTTP requests (GET, POST, PUT, DELETE, etc.).

Makes working with APIs super simple compared to low-level urllib.

🔹 Basic Example (GET Request)
import requests

# Send a GET request to a website
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

# Print the response content
print(response.text)        # raw text
print(response.json())      # parse JSON response into dict
print(response.status_code) # status code (200 = OK)

🔹 POST Request (sending data)
import requests

data = {"title": "My Post", "body": "Hello World!", "userId": 1}

response = requests.post("https://jsonplaceholder.typicode.com/posts", json=data)

print(response.json())  # shows server's response

🔹 Useful Features

.get() → fetch data

.post() → send data

.put() / .delete() → update/delete

.json() → parse JSON responses

.headers, .cookies, .status_code → inspect response

✅ Summary:
The requests module = super-easy way to talk to websites/APIs.

🔹 What is sys?

sys = System-specific parameters and functions

Built-in module (no installation needed).

Lets you interact with the Python interpreter (not the OS like os).

🔹 Common Uses of sys
1. Command-Line Arguments
import sys

# sys.argv holds arguments passed to script
print("Arguments:", sys.argv)


👉 If you run this in terminal:

python script.py hello world


Output:

Arguments: ['script.py', 'hello', 'world']

2. Exiting a Program
import sys

print("Before exit")
sys.exit()   # stops program
print("This will not print")

3. Python Version Info
import sys

print(sys.version)      # full version string
print(sys.version_info) # structured version info

4. Standard Input, Output, Error
import sys

sys.stdout.write("Hello via stdout\n")   # print alternative
sys.stderr.write("This is an error message\n")

5. Module Path (where Python looks for imports)
import sys

print(sys.path)   # list of paths Python searches for modules


✅ Summary:

os → talk to operating system (files, dirs, commands).

sys → talk to Python interpreter (args, version, exit, I/O).

```python
import re
pattern = r'\d+'
text = "There are 2 apples and 5 bananas."
matches_2=re.search(pattern, text)
matches = re.findall(pattern, text)

print(matches_2.group())  # Output: '2'
print(matches)  # Output: ['2', '5']
print(type(matches))  # Output: <class 'list'>
```
